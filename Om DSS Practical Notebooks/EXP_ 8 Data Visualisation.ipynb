```python
#importing the basic library
import numpy as np
from matplotlib import pyplot as plt
```

```python
x=np.arange(1,11)
```

```python
x
```

```python
y=2*x
```

```python
y
```

```python
plt.plot(x,y)
plt.title("Line chart")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.show()
```

```python
plt.bar(x,y)
plt.title("Bar chart")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.show()
```

```python
# scatter plot
a = (5,4,2,9,1,8,2,1,8,5)
b = (10,52,45,33,36,25,45,24,14,28)
plt.scatter(a,b)
plt.title('scatter chart')
plt.xlabel("a axis")
plt.ylabel("b axis")
plt.show()
```

```python
plt.scatter(x,y)
plt.title('scatter chart')
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.show()
```
